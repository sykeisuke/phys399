# Exercise 2: Control Flow - if, elif, else

**Goals**
- Understand the concept of control flow in programming
- Learn how to use `if`, `elif`, `else` statements for decision making

```python
# --- 1. Basic if statement ---
x = 10
if x > 5:
    print("x is greater than 5")
```

```python
# --- 2. if-else ---
score = 75
if score >= 60:
    print("Pass")
else:
    print("Fail")
```

```python
# --- 3. if-elif-else ---
energy = 15
if energy > 20:
    print("High energy")
elif energy > 10:
    print("Medium energy")
else:
    print("Low energy")
```

```python
# --- 4. Logical operators ---
a = 3
b = 7
if a > 0 and b > 0:
    print("Both are positive")

if a > 0 or b > 0:
    print("At least one is positive")

if not (a > 5):
    print("a is not greater than 5")
```

```python
# --- 5. Mini Exercise ---
# Particle identification by Time-of-Flight (TOF)
# Assume a particle travels distance L in time t
# Calculate beta = v/c and compare with expected values

L = 3.0   # meters
t_ns = 12.0   # nanoseconds
c = 299792458  # speed of light in m/s

t = t_ns * 1e-9  # convert ns to s
v = L / t
beta = v / c
print(f"Measured velocity: {v:.3e} m/s")
print(f"beta = v/c: {beta:.3f}")
```

```python
# Toy classification based on beta for particles with similar momentum
if beta > 0.95:
    print("Likely electron or another highly relativistic particle")
elif beta > 0.80:
    print("Likely pion")
elif beta > 0.60:
    print("Likely kaon")
else:
    print("Likely proton")
```
