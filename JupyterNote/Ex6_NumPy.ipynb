# Exercise 6: NumPy Basics

**Goals**
- Understand what NumPy is and why it is widely used in scientific computing
- Learn how to create and operate arrays (1D, 2D): addition, subtraction, multiplication, division
- Explore useful NumPy functions like `np.linspace`, `np.arange`, `np.zeros`, `np.ones`

```python
import numpy as np

# --- 1. Creating arrays ---
arr1 = np.array([1, 2, 3, 4, 5])
print("Array:", arr1)

arr2 = np.zeros(5)
print("Zeros:", arr2)

arr3 = np.ones((2, 3))
print("Ones (2x3):\n", arr3)

arr4 = np.arange(0, 10, 2)
print("Arange 0-10 step 2:", arr4)

arr5 = np.linspace(0, 1, 5)
print("Linspace 0-1 with 5 points:", arr5)
```

```python
# --- 2. Indexing and slicing ---
data = np.array([10, 20, 30, 40, 50])
print("data[0]:", data[0])
print("data[1:4]:", data[1:4])
print("data[-1]:", data[-1])
```

```python
# --- 3. Array operations ---
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a+b:", a + b)
print("a*b:", a * b)
print("a**2:", a**2)

# dot product
print("dot(a,b):", np.dot(a,b)) # 1*4+2*5+3*6 = 32

# cross product
print("cross(a, b):", np.cross(a, b)) 
```

```python
# --- 4. Column vectors (3x1) ---
a_col = np.array([[1], [2], [3]])
b_col = np.array([[4], [5], [6]])
print("Column vector a:\n", a_col)
print("Column vector b:\n", b_col)
print("a_col + b_col:\n", a_col + b_col)
print("a_col * b_col:\n", a_col * b_col)
```

```python
# --- 5. 2D arrays ---
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Matrix:\n", matrix)
print("Shape:", matrix.shape)
print("matrix[0,1]:", matrix[0,1])

# Multiplication
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("A @ B (matrix product):\n", A @ B)
```

```python
# --- 6. Useful functions ---
arr1 = np.array([1, 2, 3, 4, 5])
print("Mean of arr1:", np.mean(arr1))
print("Standard deviation of arr1:", np.std(arr1))
print("Max of arr1:", np.max(arr1))
```

```python
# --- 7. Mini Exercise ---
# Simulate 1000 measurements of a Gaussian-distributed variable
# with mean=0 and sigma=1

samples = np.random.normal(0, 1, 1000)
print("First 10 samples:", samples[:10])

print("Mean:", np.mean(samples))
print("Std Dev:", np.std(samples))
```
